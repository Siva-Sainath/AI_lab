```python
def draw(loc, rooms):
    a = f"[{rooms['A']}]"
    b = f"[{rooms['B']}]"
    v_a = " [V]" if loc == 'A' else "    "
    v_b = " [V]" if loc == 'B' else "    "
    print(f"\n Room A  ---  Room B")
    print(f"   {a}  -------  {b}")
    print(f"  {v_a}            {v_b}")

def run():
    rooms = {'A': 'Dirty', 'B': 'Dirty'}
    loc = 'A'
    
    while rooms['A'] != 'Clean' or rooms['B'] != 'Clean':
        draw(loc, rooms)
        if rooms[loc] == 'Dirty':
            rooms[loc] = 'Clean'
        else:
            loc = 'B' if loc == 'A' else 'A'
            
    draw(loc, rooms)
    print("Both rooms are clean!")

run()
```
