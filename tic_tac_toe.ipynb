```python
import numpy as np

def print_board(board):
    symbols = [' ', 'X', 'O']
    for i in range(0, 9, 3):
        print(f" {symbols[board[i]]} | {symbols[board[i+1]]} | {symbols[board[i+2]]} ")
        if i < 6:
            print("-----------")

def is_winner(board, player):
    wins = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    return any(all(board[i] == player for i in line) for line in wins)

def get_scores(board, player):
    scores = np.zeros(9)
    for i in range(9):
        if board[i] == 0:
            test = board.copy()
            test[i] = player
            if is_winner(test, player):
                scores[i] = 100
            elif i == 4:
                scores[i] = 3
            elif i in [0, 2, 6, 8]:
                scores[i] = 2
            else:
                scores[i] = 1
    return scores

def agent_move(board):
    scores = get_scores(board, -1)
    empty = [i for i in range(9) if board[i] == 0]
    empty_scores = scores[empty]
    probabilities = np.exp(empty_scores) / np.sum(np.exp(empty_scores))
    return np.random.choice(empty, p=probabilities)

board = np.zeros(9, dtype=int)
board[0] = 1
board[1] = -1
board[2] = 1
board[3] = -1
current_player = 1
path_cost = 0

while True:
    print_board(board)
    
    if current_player == 1:
        try:
            x, y = map(int, input("Your move (x, y): ").split(','))
            pos = x * 3 + y
            if pos < 0 or pos > 8:
                print("Invalid coordinates! Out of bounds.")
                continue
        except (ValueError, IndexError):
            print("Invalid input format! Use 'x, y' (e.g., 2,2).")
            continue

        if board[pos] != 0:
            print("Invalid move! Cell already taken.")
            continue

        board[pos] = 1
        path_cost += 1
    else:
        pos = agent_move(board)
        board[pos] = -1
        print(f"Agent chose position {pos // 3}, {pos % 3}")
        path_cost += 1

    if is_winner(board, current_player):
        print_board(board)
        print(f"Player {'X' if current_player == 1 else 'O'} wins!")
        print(f"Path cost: {path_cost}")
        break

    if all(board != 0):
        print_board(board)
        print("Draw!")
        print(f"Path cost: {path_cost}")
        break

    current_player *= -1
```
